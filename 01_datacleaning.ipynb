```python
import pandas as pd
import numpy as np

# 1. LOAD ALL 5 DATASETS
beneficiaries = pd.read_csv('CFK2_Beneficiaries.csv')
health_visits = pd.read_csv('CFK2_HealthVisits.csv')
activities    = pd.read_csv('CFK2_ProgramActivities.csv')
loans         = pd.read_csv('CFK2_SACCOLoans.csv')
budgets       = pd.read_csv('CFK2_ProjectBudgets.csv')


# ==========================================
# 2. HELPER FUNCTIONS
# ==========================================
def clean_text(df, columns):
    for col in columns:
        if col in df.columns:
            df[col] = df[col].str.replace('&nbsp;', ' ', regex=True).str.strip().str.title()
    return df

def clean_currency_col(val):
    if pd.isna(val):
        return np.nan
    if isinstance(val, str):
        val = val.replace('KES', '').replace(',', '').strip()
    try:
        return float(val)
    except:
        return np.nan

def clean_phone_col(val):
    if pd.isna(val):
        return np.nan
    return str(val).replace(' ', '').replace('-', '').strip()

def clean_fiscal_year(val):
    if pd.isna(val):
        return np.nan
    val = str(val).strip().upper()
    val = val.replace('FY', '')
    return val


# ==========================================
# 3. CLEANING BENEFICIARIES DATASET
# ==========================================
gender_map = {'m': 'Male', 'M': 'Male', 'male': 'Male', 'Male': 'Male',
              'f': 'Female', 'F': 'Female', 'female': 'Female', 'FEMALE': 'Female'}
beneficiaries['Gender'] = beneficiaries['Gender'].map(gender_map)
beneficiaries = clean_text(beneficiaries, ['Village', 'ProgramArea', 'Status'])
beneficiaries['RegistrationDate'] = pd.to_datetime(beneficiaries['RegistrationDate'], errors='coerce', format='mixed')

beneficiaries['Phone'] = beneficiaries['Phone'].apply(clean_phone_col)
beneficiaries['EmergencyContact'] = beneficiaries['EmergencyContact'].apply(clean_phone_col)


# ==========================================
# 4. CLEANING HEALTH VISITS DATASET
# ==========================================
health_visits = clean_text(health_visits, ['Service', 'Staff', 'Notes'])
health_visits['Diagnosis'] = health_visits['Diagnosis'].str.strip().str.upper()
health_visits['VisitDate'] = pd.to_datetime(health_visits['VisitDate'], errors='coerce', format='mixed')

for col in ['ConsultationFee_KES', 'MedicationCost_KES', 'TotalBill_KES']:
    health_visits[col] = health_visits[col].apply(clean_currency_col).fillna(0)

bool_map = {
    'YES': True, 'Yes': True, 'yes': True, 'Y': True, 'True': True, '1': True, 1: True,
    'NO': False, 'No': False, 'no': False, 'N': False, 'False': False, '0': False, 0: False
}
health_visits['InsuranceCovered'] = health_visits['InsuranceCovered'].map(bool_map)
health_visits['FollowUpRequired'] = health_visits['FollowUpRequired'].map(bool_map)


# ==========================================
# 5. CLEANING PROGRAM ACTIVITIES DATASET
# ==========================================
activities = clean_text(activities, ['Program', 'ActivityName', 'Attendance', 'CompletionStatus', 'Facilitator'])
activities['ActivityDate'] = pd.to_datetime(activities['ActivityDate'], errors='coerce', format='mixed')
activities['VenueCost_KES'] = activities['VenueCost_KES'].apply(clean_currency_col).fillna(0)

score_map = {
    'A': 'A', 'A-': 'A-', 'B+': 'B+', 'B': 'B', 'C+': 'C+', 'C': 'C',
    'F': 'F', 'Fail': 'F', 'Pass': 'Pass'
}
activities['ScoreGrade'] = activities['ScoreGrade'].map(score_map)
attendance_map = {
    'Present': 'Present', 'present': 'Present', 'PRESENT': 'Present', 'Present ': 'Present', '1': 'Present', 1: 'Present',
    'Absent': 'Absent', 'absent': 'Absent', 'ABSENT': 'Absent', '0': 'Absent', 0: 'Absent'
}

activities['Attendance'] = activities['Attendance'].map(attendance_map)


# ==========================================
# 6. CLEANING SACCO LOANS DATASET
# ==========================================
loans = clean_text(loans, ['Purpose', 'Status', 'OfficerName'])
loans['ApplicationDate'] = pd.to_datetime(loans['ApplicationDate'], errors='coerce', format='mixed')
loans['DisbursementDate'] = pd.to_datetime(loans['DisbursementDate'], errors='coerce', format='mixed')

for col in ['LoanAmount_KES', 'MonthlyInstalment_KES', 'TotalRepaid_KES', 'OutstandingBalance_KES']:
    loans[col] = loans[col].apply(clean_currency_col)


# ==========================================
# 7. CLEANING PROJECT BUDGETS DATASET
# ==========================================
budgets = clean_text(budgets, ['ProjectName', 'BudgetLine', 'VarianceExplained', 'ApprovedBy', 'AuditStatus'])
budgets['Donor'] = budgets['Donor'].str.strip().str.upper()
budgets['ReportingDate'] = pd.to_datetime(budgets['ReportingDate'], errors='coerce', format='mixed')

budgets['AllocatedBudget_KES'] = budgets['AllocatedBudget_KES'].apply(clean_currency_col)
budgets['ActualSpend_KES'] = budgets['ActualSpend_KES'].apply(clean_currency_col)
budgets['BudgetUtilisation_pct'] = pd.to_numeric(budgets['BudgetUtilisation_pct'], errors='coerce')
budgets['FiscalYear'] = budgets['FiscalYear'].apply(clean_fiscal_year)


# ==========================================
# 8. ADD RECORD STATUS FLAG (LOOP)
# ==========================================
registered_ids = set(beneficiaries['BeneficiaryID'])
operational_tables = {
    'Health Visits': health_visits,
    'Program Activities': activities,
    'SACCO Loans': loans
}

for name, df in operational_tables.items():
    df['RecordStatus'] = df['BeneficiaryID'].apply(
        lambda x: 'Registered Member' if x in registered_ids else 'Unregistered / Walk-in'
    )



print(" ALL DATASETS CLEANED, FLAGGED, AND SAVED SUCCESSFULLY!")
```
